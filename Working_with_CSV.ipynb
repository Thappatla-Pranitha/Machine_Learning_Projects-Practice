# OPENING A CSV FILE FROM AN URL


```python
import requests
from io import StringIO
url="https://raw.githubusercontent.com/cs109/2014_data/master/countries.csv"
headers={"User-Agent": "Mozilla/5.0 (Macintosh; Intel Mac OS X 10.14; rv:66.0) Gecko/20100101 Firefox/66.0"}
req = requests.get(url, headers=headers)
data=StringIO(req.text)
pd.read_csv(data)
```

# SEP PARAMETER

```python
pd.read_csv('orders.csv',sep=',',names=['Order_ID','Product_ID','user_ID'])
```

## INDEX_COL PARAMETER

```python
pd.read_csv('orders.csv',index_col=0)
```

## HEADER PARAMETER

```python
pd.read_csv('orders.csv',header=1)
```

## USE_COLS PARAMETER

```python
pd.read_csv('orders.csv',usecols=['User_ID','Product_ID'])
```

## ENCODING PARAMETER

```python
pd.read_csv('orders.csv',encoding='latin-1')
```

## CONVERTORS


```python
def rename(value):
    return value.replace('User','Customer')
```

```python
pd.read_csv('orders.csv',converters={'User_ID':rename})
```

## Loading a huge dataset in chunks

```python
dfs=pd.read_csv('orders.csv',chunksize=5000)
```

```python
for chunk in dfs:
    print(chunk.shape)
```
