```python
import pandas as pd
import numpy as np
import requests
from bs4 import BeautifulSoup
import time
```

```python
url = 'https://www.ambitionbox.com/list-of-companies'

webpage = requests.get(
    url,
    headers={'User-Agent': 'Mozilla/5.0'}
)

webpage.status_code
```

```python
soup = BeautifulSoup(webpage.text, 'lxml')

print(soup.prettify()[:2000])
```

```python
company = soup.find_all(
    'div',
    class_='company-content-wrapper'
)

len(company)
```

```python
name = []

for i in company:
    name.append(i.find('h2').get_text(strip=True)
                if i.find('h2') else np.nan)

name[:5]
```

```python
rating = []

for i in company:
    element = i.find('p', class_='rating')
    rating.append(element.get_text(strip=True)
                  if element else np.nan)

rating[:5]
```

```python
reviews = []

for i in company:
    element = i.find('a', class_='review-count')
    reviews.append(element.get_text(strip=True)
                   if element else np.nan)

reviews[:5]
```
